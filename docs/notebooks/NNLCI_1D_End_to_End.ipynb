# 1D NNLCI：Sod、Lax、Woodward–Colella 端到端流程

本 Notebook 用于向人或另一个 AI 说明并执行三个 1D NNLCI problem：

- Sod shock tube
- Lax shock tube（普通 MLP 和 residual-learning ResNet）
- Woodward–Colella（WC）blast-wave problem

共同思想：用两个低分辨率 CFD 场的局部窗口作为输入，预测高分辨率场的 density、velocity、pressure。

Sod/Lax 的链路为：`C++ CFD → 50/100/800 网格 .dat → NNLCI training → .pth → prediction/metrics/plots`。

WC 的链路为：`仓库已有 100/200/HF .npy → NNLCI training → .pth → prediction/plot`。当前仓库没有提交 WC 原始 CFD 批量生成器。

````python
from pathlib import Path
import os, sys, subprocess

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / 'problems').is_dir() and (candidate / 'README.md').is_file():
            return candidate
    raise FileNotFoundError('找不到 NNLCI 仓库根目录')

REPO = find_repo_root()
SOD_DIR = REPO / 'problems' / '1d_sod'
LAX_DIR = REPO / 'problems' / '1d_lax'
WC_DIR = REPO / 'problems' / '1d_woodward_colella'
print('Repository:', REPO)
print('Python:', sys.executable)
````

### WSL / Linux environment

Run this notebook from the repository root in WSL. Paths are resolved from the new `problems/` layout.

````python
import numpy as np
import pandas as pd
import scipy
import torch

print('NumPy:', np.__version__)
print('Pandas:', pd.__version__)
print('SciPy:', scipy.__version__)
print('PyTorch:', torch.__version__)
print('CUDA:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
````

## 1. Sod CFD data generation

Set `RUN_SOD_CFD = True` to compile and run the Sod solver directly inside WSL.

````python
RUN_SOD_CFD = False

if RUN_SOD_CFD:
    subprocess.run(
        [sys.executable, str(SOD_DIR / 'src' / 'cfd' / 'generate_data.py')],
        cwd=SOD_DIR / 'src' / 'cfd',
        check=True,
    )
else:
    print('Skipping Sod CFD generation; using existing data/raw files.')
````

## A2. Sod 训练

执行：`src/training/train.py`。

读取三种网格的 `train_P*.dat`。将 50/100 网格插值到 800，进行 Z-score 标准化，提取双 LF 三点窗口，因此输入为 `2×3×3=18`，输出为三个 HF 变量。网络为 8×400 GELU MLP。

默认正式训练 4000 epoch。`NNLCI_EPOCHS` 和 `NNLCI_MODEL_FILENAME` 可用于安全短测。

````python
RUN_SOD_TRAIN = False
if RUN_SOD_TRAIN:
    env = os.environ.copy()
    env['NNLCI_EPOCHS'] = '1'
    env['NNLCI_MODEL_FILENAME'] = 'smoke_model_sod.pth'
    subprocess.run([sys.executable, str(SOD_DIR / 'src' / 'training' / 'train.py')], cwd=SOD_DIR, env=env, check=True)
print('Formal model exists:', (SOD_DIR / 'models' / 'best_model_sod_final.pth').exists())
````

## A3. Sod 预测

读取 `best_model_sod_final.pth` 和 `pred_P*_{50,100,800}.dat`；生成 `results_final/*.png` 和 `sod_prediction_metrics.csv`。仓库已有模型实测全局平均相对 L2 约 1.0137%。

````python
RUN_SOD_PREDICT = False
if RUN_SOD_PREDICT:
    env = os.environ.copy(); env['MPLBACKEND'] = 'Agg'
    subprocess.run([sys.executable, str(SOD_DIR / 'src' / 'prediction' / 'predict.py')], cwd=SOD_DIR, env=env, check=True)
````

## 5. Lax CFD data generation

Set `RUN_LAX_CFD = True` to compile and run the Lax solver directly inside WSL.

````python
RUN_LAX_CFD = False

if RUN_LAX_CFD:
    subprocess.run(
        [sys.executable, str(LAX_DIR / 'src' / 'cfd' / 'generate_data.py')],
        cwd=LAX_DIR / 'src' / 'cfd',
        check=True,
    )
else:
    print('Skipping Lax CFD generation; using existing data/raw files.')
````

## B2. Lax MLP 训练与预测

普通模型使用双 LF 五点窗口，所以输入维度为 `2×5×3=30`，输出为 3。训练脚本默认 5000 epoch；预测生成 `results_lax_win5/*.png`。已有模型实测全局平均相对 L2 约 2.0318%。

````python
RUN_LAX_MLP_TRAIN = False
RUN_LAX_MLP_PREDICT = False
if RUN_LAX_MLP_TRAIN:
    env = os.environ.copy()
    env['NNLCI_EPOCHS'] = '1'
    env['NNLCI_MODEL_FILENAME'] = 'smoke_model_lax.pth'
    subprocess.run([sys.executable, str(LAX_DIR / 'src' / 'training' / 'train_mlp.py')], cwd=LAX_DIR, env=env, check=True)
if RUN_LAX_MLP_PREDICT:
    env = os.environ.copy(); env['MPLBACKEND'] = 'Agg'
    subprocess.run([sys.executable, str(LAX_DIR / 'src' / 'prediction' / 'predict_mlp.py')], cwd=LAX_DIR, env=env, check=True)
````

## B3. Lax residual-learning ResNet

ResNet 预测 HF 与插值 LF1 的 residual，而非直接预测全部场。训练脚本默认 3000 epoch。预测脚本现在会在权重缺失时明确报错，不会用随机网络输出结果。

````python
RUN_LAX_RESNET_TRAIN = False
RUN_LAX_RESNET_PREDICT = False
resnet_env = os.environ.copy()
resnet_env['NNLCI_EPOCHS'] = '1'
resnet_env['NNLCI_MODEL_FILENAME'] = 'smoke_model_lax_resnet.pth'
resnet_env['NNLCI_RESULTS_DIR'] = 'smoke_results_lax_resnet'
resnet_env['MPLBACKEND'] = 'Agg'
if RUN_LAX_RESNET_TRAIN:
    subprocess.run([sys.executable, str(LAX_DIR / 'src' / 'training' / 'train_resnet.py')], cwd=LAX_DIR, env=resnet_env, check=True)
if RUN_LAX_RESNET_PREDICT:
    subprocess.run([sys.executable, str(LAX_DIR / 'src' / 'prediction' / 'predict_resnet.py')], cwd=LAX_DIR, env=resnet_env, check=True)
````

# Part C — 1D Woodward–Colella problem

仓库包含 `training_dataset_100.npy`、`training_dataset_200.npy`、`training_dataset_hf.npy` 以及 testing 数据。当前没有用于重新生成这些原始 `.npy` 的 WC CFD 源码。

旧 `wc_nnlci.py` 原本是不可运行的 Colab 导出；现在它是本地兼容 CLI，内部调用维护中的 train/predict 脚本。

训练将 100/200 输入插值到目标网格、应用 log shift、提取局部窗口，输入为 18，输出为 density/velocity/pressure。默认正式训练 10000 epoch。

````python
wc_data_dir = WC_DIR / 'data' / 'raw'
required_wc = [
    'training_dataset_100.npy', 'training_dataset_200.npy', 'training_dataset_hf.npy',
    'testing_dataset_100.npy', 'testing_dataset_200.npy', 'testing_dataset_hf.npy'
]
for name in required_wc:
    path = wc_data_dir / name
    print(name, 'FOUND' if path.exists() else 'MISSING')
````

````python
RUN_WC_TRAIN = False
RUN_WC_PREDICT = False
if RUN_WC_TRAIN:
    subprocess.run([sys.executable, str(WC_DIR / 'src' / 'run.py'), 'train', '--epochs', '1', '--model', 'smoke_model_wc.pth'], cwd=WC_DIR, check=True)
if RUN_WC_PREDICT:
    env = os.environ.copy(); env['MPLBACKEND'] = 'Agg'
    subprocess.run([sys.executable, str(WC_DIR / 'src' / 'run.py'), 'predict', '--model', 'best_model_800.pth'], cwd=WC_DIR, env=env, check=True)
````

WC 已有正式模型 Case 55 实测平均相对 L2 约 0.5054%，并生成 `prediction_case_55.png`。
