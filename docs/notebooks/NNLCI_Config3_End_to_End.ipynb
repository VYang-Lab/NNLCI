# NNLCI Config3：端到端运行说明与可执行流程

本 Notebook 用于把 2D Riemann Config3 的完整数据链交给人或其他 AI：

`Config3 初值 → CFD 场 → 三分辨率数据 → preprocessing → 训练数组 → training → 模型权重 → post-processing → 预测与指标`

仓库主要目录：

- `src/`：2D CUDA Fortran WENO-5 CFD 求解器、Config3/4/6 namelist 生成器。
- `NNLCI/`：2D preprocessing、training、post-processing。
- `problems/1d_sod`, `problems/1d_lax`, and `problems/1d_woodward_colella`：Sod、Lax、Woodward–Colella 1D 问题，本 Notebook 不执行这些流程。

> 重要：`prepare_config3_bootstrap.py` 从仓库已有的标准 801×801 Config3 场抽取低分辨率节点，只用于验证代码链能够端到端执行。正式科研数据应对每个扰动 case 分别运行 100/200/800 分辨率 CFD。

````python
from pathlib import Path
import os, sys, subprocess

def find_repo_root(start=Path.cwd()):
    for candidate in [start, *start.parents]:
        if (candidate / 'problems' / '2d_riemann').is_dir():
            return candidate
    raise FileNotFoundError('找不到同时包含 NNLCI/ 和 src/ 的仓库根目录')

REPO = find_repo_root()
PROBLEM_DIR = REPO / 'problems' / '2d_riemann'
NNLCI_DIR = PROBLEM_DIR
SRC_DIR = PROBLEM_DIR / 'src' / 'cfd'
print('Repository:', REPO)
print('Python:', sys.executable)
````

### WSL / Linux environment

Run this notebook from the repository root in WSL. Paths are resolved from `problems/2d_riemann/`.

````python
import numpy as np
import torch
import skimage

print('NumPy:', np.__version__)
print('PyTorch:', torch.__version__)
print('scikit-image:', skimage.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('GPU:', torch.cuda.get_device_name(0))
````

## 1. 生成 Config3 namelist

执行：`problems/2d_riemann/src/cfd/generate_configs.py`

读取：脚本内部 `RIEMANN_CONFIGS`。Config3 四象限状态为 `[density, u, v, pressure]`：

- Q1 `[1.5, 0, 0, 1.5]`
- Q2 `[0.5323, 1.206, 0, 0.3]`
- Q3 `[0.138, 1.206, 1.206, 0.029]`
- Q4 `[0.5323, 0, 1.206, 0.3]`

生成：`problems/2d_riemann/configs/input_Config3_std.nml`，以及 Config4/6 namelist。

````python
RUN_CONFIG_GENERATION = False

if RUN_CONFIG_GENERATION:
    subprocess.run(
        [sys.executable, str(PROBLEM_DIR / 'src' / 'cfd' / 'generate_configs.py')],
        cwd=PROBLEM_DIR / 'src' / 'cfd',
        check=True,
    )
else:
    print('Skipping config generation; using existing configs.')
````

## Compile and run the 2D CFD solver in WSL

The CUDA Fortran solver requires NVIDIA HPC SDK (`nvfortran`) and a compatible NVIDIA GPU/driver.

```bash
cd problems/2d_riemann/src/cfd
mkdir -p bin
nvfortran -O3 -cuda weno_gpu_solver.f90 -o bin/weno_gpu_solver
bin/weno_gpu_solver ../../configs/input_Config3_std.nml ../../data/raw/flow_standard_config3.dat
```

````python
standard_flow = PROBLEM_DIR / 'data' / 'raw' / 'flow_standard_config3.dat'
assert standard_flow.exists(), f'缺少 {standard_flow}'
print('Standard Config3 CFD result:', standard_flow)
print('Size (MB):', standard_flow.stat().st_size / 1024**2)
print('Header:')
with standard_flow.open(encoding='utf-8') as stream:
    print(''.join(stream.readline() for _ in range(4)))
````

## 3. 构造 preprocessing 所需目录

执行：`problems/2d_riemann/src/preprocessing/prepare_config3_bootstrap.py`。

读取：`problems/2d_riemann/data/raw/flow_standard_config3.dat`。

生成：

- `problems/2d_riemann/data/bootstrap/DATA_TRAIN/Config3/case_000/flow_100x100.dat`
- `problems/2d_riemann/data/bootstrap/DATA_TRAIN/Config3/case_000/flow_200x200.dat`
- `problems/2d_riemann/data/bootstrap/DATA_TRAIN/Config3/case_000/flow_800x800.dat`

Tecplot 文件实际含边界节点，因此对应节点数组为 101×101、201×201、801×801；preprocessing 会裁成目标分辨率。

````python
subprocess.run([sys.executable, str(PROBLEM_DIR / 'src' / 'preprocessing' / 'prepare_config3_bootstrap.py')], cwd=NNLCI_DIR, check=True)
case_dir = PROBLEM_DIR / 'data' / 'bootstrap' / 'DATA_TRAIN' / 'Config3' / 'case_000'
for path in sorted(case_dir.glob('*.dat')):
    print(path.name, f'{path.stat().st_size / 1024**2:.2f} MB')
````

## 4. Preprocessing

执行：`preprocess.py`。

读取：`DATA_TRAIN/Config3/case_*/flow_{100,200,800}x{100,200,800}.dat`。

处理：LF 双线性上采样到 800；基于 HF 梯度保留高梯度 40% 和部分平滑区；提取两套 3×3、dilation=4 stencil；拼成 72 维输入；目标是 HF 的 `D,U,V,P`；输入缩放至 `[-1,1]`。

生成：`train_input_config3.npy`、`train_target_config3.npy`、`x_scaler_config3.npz`。

````python
subprocess.run([sys.executable, str(PROBLEM_DIR / 'src' / 'preprocessing' / 'preprocess.py')], cwd=NNLCI_DIR, check=True)
stencil_dir = PROBLEM_DIR / 'data' / 'processed' / 'NNLCI_Data' / 'Config3_Stencils'
x = np.load(stencil_dir / 'train_input_config3.npy', mmap_mode='r')
y = np.load(stencil_dir / 'train_target_config3.npy', mmap_mode='r')
print('train input:', x.shape, x.dtype)
print('train target:', y.shape, y.dtype)
````

## 5. Training

读取：preprocessing 生成的两个 `.npy`。模型是 `72 → 10×600 Tanh → 4` 的 MLP，使用 AdamW、MSE 和 ReduceLROnPlateau。

生成：`problems/2d_riemann/models/NNLCI_Models/nnlci_config3_deep_latest.pth`。

下面默认只训练 2 epoch 验证链路。正式训练把 `NNLCI_EPOCHS` 改为 `500`。

````python
train_env = os.environ.copy()
train_env['NNLCI_EPOCHS'] = '2'       # 正式训练改成 500
train_env['NNLCI_BATCH_SIZE'] = '65536'
train_env['NNLCI_NUM_WORKERS'] = '0'  # Jupyter 稳定设置
subprocess.run([sys.executable, str(PROBLEM_DIR / 'src' / 'training' / 'train.py')], cwd=NNLCI_DIR, env=train_env, check=True)
model_path = PROBLEM_DIR / 'models' / 'NNLCI_Models' / 'nnlci_config3_deep_latest.pth'
print('Model:', model_path, f'{model_path.stat().st_size / 1024**2:.2f} MB')
````

## 6. Post-processing

读取：模型权重，以及 `train_input_config3.npy`、`train_target_config3.npy`。当前代码用训练数据演示评估；正式实验应另建 test 数据。

输出指标：MAE、MSE、Relative L1/L2、PSNR；`GRID_SIZE=None` 时跳过 SSIM。

生成：`problems/2d_riemann/results/NNLCI_Output/predictions_config3.npy`。

````python
eval_env = os.environ.copy()
eval_env['NNLCI_NUM_WORKERS'] = '0'
subprocess.run([sys.executable, str(PROBLEM_DIR / 'src' / 'prediction' / 'evaluate.py')], cwd=NNLCI_DIR, env=eval_env, check=True)
prediction_path = PROBLEM_DIR / 'results' / 'NNLCI_Output' / 'predictions_config3.npy'
pred = np.load(prediction_path, mmap_mode='r')
print('Predictions:', pred.shape, pred.dtype)
print('Saved at:', prediction_path)
````
